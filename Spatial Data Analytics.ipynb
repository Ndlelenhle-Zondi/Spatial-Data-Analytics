```python
import geopandas as gpd
from shapely.geometry import Point

data = {
    "City": ["Cape Town", "Johannesburg"],
    "Latitude": [-33.9249, -26.2041],
    "Longitude": [18.4241, 28.0473]
}

geometry = [
    Point(18.4241, -33.9249),
    Point(28.0473, -26.2041)
]

gdf = gpd.GeoDataFrame(data, geometry=geometry, crs="EPSG:4326")

gdf
```

```python
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="my_database",
    user="postgres",
    password="your_password"
)

cursor = conn.cursor()

cursor.execute("""
    SELECT *
    FROM customers
    LIMIT 10;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)

cursor.close()
conn.close()
```
